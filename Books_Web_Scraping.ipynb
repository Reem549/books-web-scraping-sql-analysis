# Books Web Scraping

## Project Overview

This project collects data from the first 100 books on [Books to Scrape](https://books.toscrape.com/) and stores the results in a CSV file for further analysis using SQL.

### Objectives

- Scrape the first 100 books from pages 1 to 5.
- Extract the book title, price, rating, stock availability, and URL.
- Convert price and rating into the required data types.
- Save the scraped data into `books.csv`.


```python
# import libraries
import requests
import csv
from urllib.parse import urljoin
from bs4 import BeautifulSoup
import pandas as pd
```

```python
#get URL and request
url="https://books.toscrape.com/"
response= requests.get(url)
if response.status_code==200:
    print("Request Successful")
else:
  print("Request Failed")
```

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup)
```

#Extract Books From Page 1


```python
from urllib.parse import urljoin

books = soup.find_all("h3")


books_extracted = 0

rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

data = []

# iterate through books and extract the information for each book
for book in books:

    book_url = book.find("a")["href"]

    # Create the full URL of the book
    book_full_url = urljoin(url, book_url)

    book_response = requests.get(book_full_url)
    book_response.encoding = "utf-8"

    book_soup = BeautifulSoup(
        book_response.text,
        "html.parser"
    )

    # Title
    title = book_soup.find("h1").text.strip()

    # Price
    price = book_soup.find(
        "p",
        class_="price_color"
    ).text.strip()

    price = float(
        price.replace("Â", "").replace("£", "")
    )

    # Rating
    rating = book_soup.find(
        "p",
        class_="star-rating"
    )["class"][1]

    rating = rating_map[rating]

    # Availability
    availability = book_soup.find(
        "p",
        class_="availability"
    ).text.strip()

    in_stock = "In stock" in availability

    # Store data
    data.append({
        "title": title,
        "price": price,
        "rating": rating,
        "in_stock": in_stock,
        "url": book_full_url
    })



    print(f"Title: {title}")
    print(f"Price: {price}")
    print(f"Rating: {rating}")
    print(f"In Stock: {in_stock}")
    print(f"URL: {book_full_url}")
```

#Extract Books From Pages 1-5


```python

rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

data = []


books_extracted = 0


# Iterate through pages 1 to 5
for page in range(1, 6):

    # Page URL
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    # Send request
    response = requests.get(url)

    # Parse HTML
    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    # Find all books
    books = soup.find_all("h3")


    # Iterate through books
    for book in books:

        # Get book relative URL
        book_url = book.find("a")["href"]

        # Create full book URL
        book_full_url = urljoin(url, book_url)

        # Request book page
        book_response = requests.get(book_full_url)

        # Parse book page
        book_soup = BeautifulSoup(
            book_response.text,
            "html.parser"
        )



        # Title


        title = book_soup.find("h1").text.strip()



        # Price


        price = book_soup.find(
            "p",
            class_="price_color"
        ).text.strip()

        price = float(
            price.replace("Â", "").replace("£", "")
        )



        # Rating


        rating = book_soup.find(
            "p",
            class_="star-rating"
        )["class"][1]

        rating = rating_map[rating]



        # Availability


        availability = book_soup.find(
            "p",
            class_="availability"
        ).text.strip()

        in_stock = "In stock" in availability



        # Store data


        data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "in_stock": in_stock,
            "url": book_full_url
        })


        books_extracted += 1


print("Books extracted:", books_extracted)
```

#Export the Data


```python
df = pd.DataFrame(data)

df.to_csv("books.csv", index=False)

df.head()
```
